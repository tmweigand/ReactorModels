```python
import reactormodels
from reactormodels.postprocess import plot_breakthrough_and_model, plot_breakthrough

import numpy as np
import copy

data = np.genfromtxt("experimental_data/in/exp31.txt", delimiter="\t")
species_list = np.genfromtxt(
    "experimental_data/in/exp31.txt", delimiter="\t", dtype=str, max_rows=1
)[1:]
species_list = np.char.strip(species_list)
idx = species_list.tolist().index("PMPA")
influent_data = data[1:3, 1:]
effluent_data = data[3:, 1:]
effluent_data[1, idx] = np.nan
bed_volumes = data[3:, 0]

# particle
particle_porosity = 0.5
particle_density = 600  # g/L
particle_diameter = 0.007 * 2  # cm

# column
length = 1.5  # cm
diameter = 0.318  # cm
porosity = 0.38
bulk_density = 399.8  # g/L
flow_rate = 7.14 / 60  # cm3/s

media = reactormodels.Media(
    particle_porosity=particle_porosity,
    particle_diameter=particle_diameter,
    particle_density=particle_density,
)

column = reactormodels.Column(
    length=length,
    porosity=porosity,
    diameter=diameter,
    bulk_density=bulk_density,
    media=media,
    water=reactormodels.Water(),
)

breakthroughs = []
for i in range(len(species_list)):
    chemical = reactormodels.Chemical(name=species_list[i])
    raw = reactormodels.Breakthrough(
        column=column,
        chemical=chemical,
        feed_concentrations=influent_data[:, i],
        flow_rate=flow_rate,
        bed_volumes=bed_volumes,
        effluent_concentrations=effluent_data[:, i],
    )
    breakthroughs.append(raw)

plot_breakthrough(
    breakthroughs, species_list, save_path=f"experimental_data/out/exp31.png"
)
```
